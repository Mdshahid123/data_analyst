## <h1 style=color:yellow>***1) pandas series:***</h1>A Pandas Series is a one-dimensional, labeled array capable of holding any data type 


```
0    2
1    3
2    4
3    5
4    6

It looks like  single column in an Excel spreadsheet or a database table.

By default, Pandas creates a numeric index starting from 0.

if we want diffrent index instead of numeric then we can give extra parameter index while creating a series 


```




````python
# creating a panads series using list 
import pandas as pd 
ser=pd.Series([2,3,4,5,6])
print(ser)
````

````python
# using array
import pandas as pd 
import array 
my_arr=array.array("i",[2,3,4,5,6])
print(pd.Series(my_arr))
````

````python
# using numpy array
import numpy as np 
import pandas as pd 

my_arr=np.array([2,3,1,8])
print(pd.Series(my_arr))
````

````python
# using dictinary 
import pandas as pd 

pd.Series({
      "name":"shahid",
      "age":24,
      "college":"IIT"
})

````

## 2) Data frame in pandas 

## Pandas DataFrame is a two-dimensional, tabular data structure
## Think of it exactly like a spreadheet / Excel sheet or a SQL database table


## default index(label) of row is numeric starting with 0 



````python
# creating dataframes using dict 

import pandas as pd

my_dic = { "name": ["shahid","raj","rakesh"], "age": [24,30,35], "sallary": [40000,30000,45000] }
df = pd.DataFrame(my_dic)
print(df)
````

````python
# creating data frames using list 
import pandas as pd
my_list=[["shahid",24,"Delhi university"],["fahad",26,"iit"],["sanajy",18,"Delhi university"]]
pd.DataFrame(my_list)
````

````python
# now we can give the custom labels to rows and columns 

pd.DataFrame(my_list,columns=["name","age","College"])
pd.DataFrame(my_list,columns=["name","age","sallery"],index=["a","b","c"])

````

# <h1 style=color:pink>***Selecting the data***</h1>

## a)Selecting Columns

```
df["column_name"]        # Single column (as Series)
df[["col1", "col2"]]     # Multiple columns (as DataFrame)

````python
import pandas as pd 

df=pd.read_csv("train.csv")

# slect the single col 

df[["Pclass"]]

# multiple col 

df[["Pclass","Survived","Name"]]

````

## b)Selecting Rows

```
Use .loc[] (label-based) and .iloc[] (position-based):

df.loc[0]                # First row (by label)
df.iloc[0]               # First row (by position)

````python
import pandas as df
df=df.read_csv("train.csv")

# select one row 
df.loc[[4]] 

# multiple rows 

df.loc[[4,3,5]]
````

## c)select a cell

````python
import pandas as pd 
df=pd.read_csv("train.csv")
df.loc[[5],"Sex"]
````

## d)slicing

````python
import pandas as pd 
df=pd.read_csv("train.csv")
df.loc[0:4,["PassengerId","Survived","Pclass","Name"]] # last value of slicinf for row will be included 
````

````python
import pandas as pd 
df=pd.read_csv("train.csv")
df.iloc[0:4,0:3] # last value of slicining  will be exluded  in case of iloc 
````

## e)Fast Access: .at and .iat
## These are optimized for single element or cell  access:

```
df.at[0, "Name"]       # Fast label-based access
df.iat[0, 1]           # Fast position-based access


````python
import pandas as pd  
df=pd.read_csv('train.csv')
df.at[1,"Pclass"]
````
