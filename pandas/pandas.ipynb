## <h1 style=color:yellow>***1) pandas series:***</h1>A Pandas Series is a one-dimensional, labeled array capable of holding any data type 


```

It looks like  single column in an Excel spreadsheet or a database table.

By default, Pandas creates a numeric index starting from 0.

if we want diffrent index instead of numeric then we can give extra parameter index while creating a series 


```




````python
# creating a panads series using list 
import pandas as pd 
ser=pd.Series([2,3,4,5,6])
print(ser)
````

````python
# using array
import pandas as pd 
import array 
my_arr=array.array("i",[2,3,4,5,6])
print(pd.Series(my_arr))
````

````python
# using numpy array
import numpy as np 
import pandas as pd 

my_arr=np.array([2,3,1,8])
print(pd.Series(my_arr))
````

# <h1 style=color:pink>***Chapter-06 Filtering the data  Based on condition***</h1>


### Data filtering ka matlab hai: DataFrame mein se sirf woh rows select karna jo hamari condition satisfy karti hain.

###  For example, agar 5 students ka data hai aur humein sirf un students ko dekhna hai jinke marks 70 se zyada hain, toh hum filtering use karenge.

## 1. Single Conditions
## Comparison operators: >, <,<=,>=, ==, !=

## ques: Hum ek students DataFrame banate hain. Isi data ko use karke saare filtering concepts seekhenge.

````python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Aman", "Sara", "Rahul", "Neha", "Imran"],
    "Age": [20, 22, 19, 24, 21],
    "Marks": [65, 85, 45, 92, 75],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"]
})

df
````

## Suppose humein check karna hai ki kaunse students ke marks 70 se zyada hain.

````python
df["Marks"]>70 
````

````python
df
````

````python
df[df["Marks"] > 70]
````

## Suppose humein 70 se zyada marks wale students ke sirf names aur marks chahiye.

````python
df[df["Marks"]>70][["Name","Marks"]]
````

# 2) Multiple Conditions in Pandas

#### &( boollean and operator) :  Dono conditions True honi chahiye 
#### |( boolean or operator ) :  Koi ek condition True honi chahiye 
#### ~ |(boolean not operator) :  Condition ka result reverse karta hai 

## a). AND (&) — Dono conditions true honi chahiye
#### Example: students jinke marks 70 se zyada hain aur city Mumbai hai.

````python
df[(df["Marks"] > 70) & (df["City"] == "Mumbai")]
````

## b). OR (|) — Koi ek condition true honi chahiye
## Example: un students ko find karo jo Delhi mein rehte hain ya jinke marks 80 se zyada hain.

````python
df[(df["City"] == "Delhi") | (df["Marks"] > 80)]
````

## c). NOT (~) — Condition ko reverse karna
## Example: un students ko find karo jo Delhi mein nahi rehte.

````python
df["City"]==
````

## 4)Text Filtering in Pandas

```
strings ya text ke basis par rows filter karenge. Jaise names mein koi word search karna, kisi prefix se start hone wale names find karna, ya kisi particular ending wale names select karna.

````python
df[df["Name"].str.contains("a", case=False)]
````

````python
df[df["Name"].str.contains("an", case=False)]
````

````python
df[df["Name"].str.startswith("a", na=False, case=False)]
````

````python
df[df["Name"].str.lower().str.startswith("a", na=False)]
````

````python
df[df["Name"].str.endswith("a", na=False)]
````

````python
df[df["Name"].str.lower().str.endswith("a", na=False)]
````
