#  ***chapter-02 : Data structure in pandas*** 

## 1) series:A Pandas Series is a one-dimensional, labeled array capable of holding any data type

coneptually it looks like a  single column in an Excel spreadsheet or a database table.

````python
# creating a panads series using list 

import pandas as pd 

series=pd.Series([2,3,4,5,6,7,8])
print(series)
````

````python
# creating pandas series using array 

import array 
import pandas as pd 


my_arr=array.array("i",[2,3,4,5,6,7,8,9])

series=pd.Series(my_arr)

print(series)
````

````python
import numpy as np
import pandas as pd

arr=np.array([2,3,4,5,6,7,8,9])

series=pd.Series(arr)
print(series)
````

````python
# creating a pandas series using dictinary 

import pandas as pd 
series=pd.Series({"name":"shahid","college":"iit","age":18})
print(series)
````

#  ***2) Data frame in pandas***

## Pandas DataFrame is a two-dimensional, tabular data structure
## Think of it exactly like a spreadheet / Excel sheet or a SQL database table


## default index(label) of row is numeric starting with 0 


````python
# create a data frame by reading a csv file 
import pandas as pd 
csv=pd.read_csv("train.csv")
df=pd.DataFrame(csv)
print(df)
````

````python
# creating a data frames using dictinary 

import pandas as pd 
df=pd.DataFrame({"name":["shahid"],"age":[18],"college":["iit"]})
print(df)
````
